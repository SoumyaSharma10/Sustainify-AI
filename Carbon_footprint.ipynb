```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Sample values
electricity_bill = 3200      # ₹
tariff_rate = 8              # ₹ per unit
water_usage = 15000          # litres/month
```

```python
electricity_units = electricity_bill / tariff_rate

print("Electricity Units:", electricity_units, "kWh")
```

```python
emission_factor = 0.82

electricity_co2 = electricity_units * emission_factor

print("Electricity CO₂:", electricity_co2, "kg/month")
```

```python
water_emission_factor = 0.0003

water_co2 = water_usage * water_emission_factor

print("Water CO₂:", water_co2, "kg/month")
```

```python
total_co2 = electricity_co2 + water_co2

print("Total Carbon Footprint:", total_co2, "kg CO₂/month")
```

```python
if total_co2 < 150:
    score = 95
    status = "Excellent"

elif total_co2 < 250:
    score = 85
    status = "Good"

elif total_co2 < 350:
    score = 75
    status = "Average"

elif total_co2 < 500:
    score = 60
    status = "Needs Improvement"

else:
    score = 40
    status = "Poor"

print("Sustainability Score:", score)
print("Status:", status)
```

```python
annual_co2 = total_co2 * 12

trees = annual_co2 / 21

print("Approximate Trees Required:", round(trees))
```

```python
print("-------- Report --------")

print("Electricity Bill : ₹", electricity_bill)

print("Water Usage :", water_usage, "litres")

print("Electricity Units :", round(electricity_units,2),"kWh")

print("Electricity CO₂ :", round(electricity_co2,2),"kg")

print("Water CO₂ :", round(water_co2,2),"kg")

print("Total CO₂ :", round(total_co2,2),"kg")

print("Sustainability Score :", score)

print("Status :", status)

print("Trees Required :", round(trees))
```

```python
categories = ['Electricity', 'Water']

values = [electricity_co2, water_co2]

plt.figure(figsize=(6,4))

plt.bar(categories, values)

plt.title("CO₂ Contribution")

plt.ylabel("kg CO₂")

plt.show()
```

```python
plt.figure(figsize=(5,5))

plt.pie(values,
        labels=categories,
        autopct="%1.1f%%")

plt.title("Carbon Footprint Distribution")

plt.show()
```

```python
plt.figure(figsize=(6,2))

plt.bar(["Score"], [score])

plt.ylim(0,100)

plt.title("Sustainability Score")

plt.show()
```
